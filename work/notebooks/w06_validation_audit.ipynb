## 1. Two paper findings + my methodology questions

### Finding 1: Growing content is younger than declining content

The paper reports that growing pages have an average content age
of 185 days, compared with 228 days for declining pages.
Their average word counts are almost identical: 1,487 and 1,481 words.

**Where does the outcome come from?**

The paper defines trend direction using the last 30 days compared
with the preceding 30 days. Growth above 10% is classified as
up, while a decline greater than 10% is classified as down.
These categories describe observed performance changes.

**My methodology question: Does the validation support the claim?**

The average-age comparison supports the observation that growing
pages are younger in this dataset. However, it does not establish
that older content causes declining performance. Other factors
could contribute to the difference, and the general study
methodology explicitly treats the findings as patterns rather
than proof of cause and effect.

I would interpret this as a useful directional signal for further
investigation, not as proof that updating an older page will
improve its performance.

### Finding 2: AI-generated content is not automatically penalized

The paper compares content associated with different AI writing
systems across age groups. The stronger-performing system varies
across cohorts rather than showing one system consistently winning.
The paper also notes that Gemini was adopted later, limiting
comparisons in older age groups.

**Where does the outcome come from?**

The analysis compares observed performance measures, including
health scores, impressions, and average search position across
AI writing-system cohorts and content-age groups. It does not
directly measure a universal Google penalty for AI-generated
content.

**My methodology question: Does the validation support the claim?**

Comparing content within age groups helps reduce differences
caused by age. However, age is not the only possible confounder.
Topic, search intent, client, editorial quality, and publishing
practices may also affect performance. Different sample sizes
and the later adoption of Gemini further limit direct comparisons.

The evidence supports the narrower conclusion that the observed
data does not show a consistent disadvantage associated with
one AI writing system across all age groups. It does not prove
that AI-generated content can never be disadvantaged or that
the same result applies to every website.

### My takeaway

Both findings provide useful evidence about patterns in the
observed dataset. Before turning either finding into a business
recommendation, I would check how the outcomes were defined,
whether the comparisons account for important confounders, and
whether the validation design supports the strength of the claim.

### Validation comparison: Random split vs. client-grouped split

My Week-5 notebook evaluated Logistic Regression, Decision Tree, and Random Forest using a client-grouped train/test split.

The ML-07 notebook used a rule-based action score to rank content pages. It did not report a train/test evaluation metric for that ranking, so I cannot directly compare an ML-07 holdout score with the Week-5 model scores.

For this audit, I will compare the same Random Forest model under two validation strategies:

1. **Random split:** Individual content pages are randomly assigned to training and test sets. Pages from the same client may appear in both sets.
2. **Client-grouped split:** Training and test sets contain different clients.

I will keep the target definition, candidate features, Random Forest configuration, test proportion, and Average Precision metric consistent across the two experiments.

This comparison will help me understand how the validation design affects measured performance. The random split is a reference comparison, while the client-grouped split tests generalization to clients not represented in training.

The target is derived from the existing `trend_direction` field. Therefore, this experiment evaluates the prediction of an observed trend label; it does not establish that the model can predict future content decline.


Relationship to my Week-5 evaluation
My original Week-5 notebook already used a client-grouped train/test split. For this audit, I added an ordinary random split as a reference and compared it with the client-grouped result. The random-split Average Precision was 0.7886, compared with 0.6109 under the client-grouped split. This comparison shows that the measured performance depends on the validation strategy. It does not represent an improvement over my original Week-5 evaluation; instead, it demonstrates why the validation design should match the intended use case.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


import pandas as pd

from sklearn.model_selection import train_test_split, GroupShuffleSplit
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import average_precision_score


# 1. Load the same dataset used in Week-5

url = (
    "https://raw.githubusercontent.com/"
    "flyrank-bih/flyrank-ml-internship-starter/"
    "main/data/raw/content_refresh_anonymized.csv"
)

df = pd.read_csv(url)

print("Original dataset shape:", df.shape)


# 2. Prepare the eligible population

eligible = df.loc[
    (df["impressions_90d"] > 0)
    & (df["content_age_days"] >= 90)
].drop_duplicates(subset="content_id").copy()

eligible["is_declining"] = (
    eligible["trend_direction"] == "down"
).astype(int)

print("Eligible pages:", len(eligible))
print("Unique clients:", eligible["client_id"].nunique())

print(
    "Declining label rate (%):",
    round(eligible["is_declining"].mean() * 100, 2)
)


# 3. Define the same candidate features as Week-5

numeric_candidates = [
    "search_volume",
    "competition",
    "cpc",
    "word_count",
    "char_count",
    "content_age_days",
    "days_since_last_update",
    "impressions_90d",
    "clicks_90d",
    "ctr",
    "avg_position",
    "sessions_90d",
    "engagement_rate",
    "scroll_rate"
]

categorical_candidates = [
    "competition_level",
    "content_type",
    "main_intent",
    "provider_used",
    "model_used"
]

numeric_features = [
    col for col in numeric_candidates
    if col in eligible.columns
]

categorical_features = [
    col for col in categorical_candidates
    if col in eligible.columns
]

feature_columns = numeric_features + categorical_features

forbidden_features = {
    "content_id",
    "client_id",
    "is_declining",
    "trend_direction",
    "trend_pct",
    "is_declining_label"
}

assert not (set(feature_columns) & forbidden_features)

X = eligible[feature_columns].copy()
y = eligible["is_declining"].copy()


# 4. Define the preprocessing pipeline

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("numeric", numeric_pipeline, numeric_features),
    ("categorical", categorical_pipeline, categorical_features)
])


# 5. Use the same Random Forest configuration as Week-5

def build_model():
    return Pipeline([
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=200,
            min_samples_leaf=10,
            random_state=42,
            n_jobs=-1
        ))
    ])


# 6. Experiment A: Ordinary random split

X_train_random, X_test_random, y_train_random, y_test_random = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=42,
        stratify=y
    )
)

random_model = build_model()

random_model.fit(X_train_random, y_train_random)

random_probabilities = random_model.predict_proba(
    X_test_random
)[:, 1]

random_ap = average_precision_score(
    y_test_random,
    random_probabilities
)


# 7. Experiment B: Client-grouped split

group_splitter = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    group_splitter.split(
        eligible,
        y=y,
        groups=eligible["client_id"]
    )
)

X_train_grouped = X.iloc[train_idx]
X_test_grouped = X.iloc[test_idx]

y_train_grouped = y.iloc[train_idx]
y_test_grouped = y.iloc[test_idx]

grouped_model = build_model()

grouped_model.fit(
    X_train_grouped,
    y_train_grouped
)

grouped_probabilities = grouped_model.predict_proba(
    X_test_grouped
)[:, 1]

grouped_ap = average_precision_score(
    y_test_grouped,
    grouped_probabilities
)


# 8. Verify that no client appears in both grouped sets

train_clients = set(
    eligible.iloc[train_idx]["client_id"]
)

test_clients = set(
    eligible.iloc[test_idx]["client_id"]
)

overlapping_clients = train_clients & test_clients

assert len(overlapping_clients) == 0


# 9. Compare the results

comparison = pd.DataFrame({
    "Validation strategy": [
        "Random split",
        "Client-grouped split"
    ],
    "Training rows": [
        len(X_train_random),
        len(X_train_grouped)
    ],
    "Test rows": [
        len(X_test_random),
        len(X_test_grouped)
    ],
    "Test decline rate (%)": [
        round(y_test_random.mean() * 100, 2),
        round(y_test_grouped.mean() * 100, 2)
    ],
    "Average Precision": [
        round(random_ap, 4),
        round(grouped_ap, 4)
    ]
})

print("\nValidation comparison:")
print(comparison.to_string(index=False))

print("\nTraining clients in grouped split:", len(train_clients))
print("Test clients in grouped split:", len(test_clients))
print("Clients appearing in both grouped sets:", len(overlapping_clients))

print(
    "\nAverage Precision difference "
    "(grouped - random):",
    round(grouped_ap - random_ap, 4)
)
```

## Results and interpretation

### Observed results

The Random Forest model achieved an Average Precision of 0.7838
under the ordinary random split and 0.5803 under the client-grouped
split.

The difference in Average Precision (grouped minus random) was
-0.2034. This is approximately 25.9% lower relative to the
random-split score.

The ordinary random split contained 24,000 training rows and 6,000
test rows. The client-grouped split contained 23,837 training rows
and 6,163 test rows.

The grouped split separated 25 training clients from 7 test clients,
with no client appearing in both sets.

### What changed?

The ordinary random split allows pages from the same client to
appear in both training and test sets. The client-grouped split
evaluates the model on clients not represented in training.

The lower Average Precision under the grouped split is consistent
with the possibility that the ordinary random split gives a more
optimistic estimate when the intended use case involves unseen
clients.

However, the two test sets contain different pages and have slightly
different declining-label rates. This experiment does not establish
that client overlap alone explains the performance difference.

### My conclusion

In this experiment, the measured Average Precision was sensitive
to the validation strategy. For a use case involving unseen clients,
I would prioritize the client-grouped evaluation over the
ordinary random-split result.

The target is derived from the existing `trend_direction` field.
Therefore, these results evaluate prediction of an observed trend
label; they do not establish that the model can predict future
content decline or that its measured performance will remain the
same in production.

## 3. Leakage audit

### Leakage audit: Are any features giving the model information about the answer?

The target variable, `is_declining`, is derived from the existing `trend_direction` field. In this dataset, `trend_direction` describes an observed content trend rather than a future outcome.

I need to check two types of potential leakage:

1. **Direct target leakage:** Features such as `trend_direction`, `trend_pct`, or `is_declining_label` could reveal the target directly or provide a duplicate of it. These should not be included as model features.
2. **Potential time-related leakage:** Metrics such as impressions, clicks, sessions, CTR, and average position may be related to the performance information used to define the observed trend. Their availability and measurement periods need to be checked before claiming that the model can predict future decline.

The code below checks whether forbidden target-related columns are present in the candidate feature set and identifies features that need further investigation.

This is a feature-level audit. It can identify suspicious columns, but it cannot establish that every remaining feature is leakage-free without confirming when each feature was measured and how the target was constructed.




### Leakage audit: Results and limitations

**Direct target leakage — Result**

The direct leakage check found no forbidden target-related columns in the final feature set. In particular, the feature set excludes `is_declining`, `trend_direction`, `trend_pct`, and `is_declining_label`.

This is a successful check for the explicitly forbidden columns, but it does not establish that every feature is safe for every prediction use case.

**Performance-related features requiring a timing review**

The audit identified seven performance-related features requiring further review: `impressions_90d`, `clicks_90d`, `sessions_90d`, `ctr`, `avg_position`, `engagement_rate`, and `scroll_rate`.

These variables may contain information related to the observed trend. Whether they constitute leakage depends on when they were measured relative to the period used to assign the target. The current audit flags this concern but does not independently establish the measurement timing of each feature.

**Target definition and limitation**

The target `is_declining` matches the condition `trend_direction == 'down'`. Therefore, the model evaluates how well these features classify an existing trend label.

Without a clearly defined future prediction period and confirmation that all features would be available before that period, the results should not be described as evidence that the model predicts future content decline.

**Audit conclusion**

No direct leakage was found among the explicitly forbidden columns. However, the timing of the seven performance-related features remains an unresolved concern. The current checks do not prove that the feature set is leakage-free for future prediction.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.


# 1. Define columns that must not be used as model features

forbidden_features = {
    "content_id",
    "client_id",
    "is_declining",
    "trend_direction",
    "trend_pct",
    "is_declining_label"
}

# 2. Check whether any forbidden columns are in our feature set

direct_leakage_candidates = (
    set(feature_columns) & forbidden_features
)

print("Direct target leakage check")
print("----------------------------")
print("Forbidden columns found in features:", direct_leakage_candidates)

if len(direct_leakage_candidates) == 0:
    print("PASS: No forbidden target-related columns are in the feature set.")
else:
    print("WARNING: Remove the forbidden columns before model training.")


# 3. Identify performance-related features that need a timing check

performance_features_to_review = [
    "impressions_90d",
    "clicks_90d",
    "sessions_90d",
    "ai_sessions_90d",
    "days_with_impressions",
    "days_with_sessions",
    "ctr",
    "avg_position",
    "engagement_rate",
    "scroll_rate",
    "ai_traffic_pct"
]

performance_features_found = [
    col for col in performance_features_to_review
    if col in feature_columns
]

print("\nPerformance-related features requiring a timing check")
print("-----------------------------------------------------")

for col in performance_features_found:
    print("-", col)


# 4. Confirm how the target was constructed

print("\nTarget definition check")
print("----------------------")
print("Target column: is_declining")
print("Target source: trend_direction")

print(
    "Target matches trend_direction == 'down':",
    (
        eligible["is_declining"]
        == (eligible["trend_direction"] == "down").astype(int)
    ).all()
)


# 5. Summarize the audit

print("\nAudit summary")
print("-------------")

if len(direct_leakage_candidates) == 0:
    print("No direct target-related leakage candidates were found.")
else:
    print("Direct leakage candidates require correction.")

print(
    "Performance-related features requiring timing review:",
    len(performance_features_found)
)

print(
    "Important limitation: this check does not prove that all features "
    "are safe for future prediction."
)
```
