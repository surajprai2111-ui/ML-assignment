## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

# Replace this with the actual starter CSV path
df = pd.read_csv("/content/content_refresh_anonymized.csv")

print("Rows:", len(df))
print("Columns:", len(df.columns))

display(df.head())
print(df.columns.tolist())
```

```python
impressions_col = "impressions_90d"
clicks_col = "clicks_90d"

total_impressions = df[impressions_col].sum()
total_clicks = df[clicks_col].sum()

overall_ctr = (total_clicks / total_impressions) * 100

print("Total impressions:", total_impressions)
print("Total clicks:", total_clicks)
print("Overall CTR:", round(overall_ctr, 2), "%")
```
